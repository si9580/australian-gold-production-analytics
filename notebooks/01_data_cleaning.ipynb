```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

file_path = "resources-and-energy-quarterly-june-2026-historical-data.xlsx"

gold_raw = pd.read_excel(
    file_path,
    sheet_name="27",
    header=None
)

gold_raw.shape
```

```python
gold_raw.head(15)
```

```python
gold_raw.iloc[7:30, 5:20]
```

```python
dates = gold_raw.iloc[7, 11:]
production = gold_raw.iloc[18, 11:]

gold_production = pd.DataFrame({
    "date": dates,
    "gold_production_tonnes": production
})

gold_production.head()
```

```python
gold_production.tail()
```

```python
gold_raw.iloc[20:48, 5:20]
```

```python
gold_df = pd.DataFrame({
    "date": gold_raw.iloc[7, 11:],
    "production_tonnes": gold_raw.iloc[18, 11:],
    "export_quantity_tonnes": gold_raw.iloc[35, 11:],
    "export_value_aud_million": gold_raw.iloc[37, 11:],
    "gold_price_usd_oz": gold_raw.iloc[42, 11:],
    "gold_price_aud_oz": gold_raw.iloc[43, 11:]
})

gold_df.head()
```

```python
gold_df.shape
```

```python
gold_df.isnull().sum()
```

```python
gold_df["date"] = pd.to_datetime(gold_df["date"])

numeric_cols = [
    "production_tonnes",
    "export_quantity_tonnes",
    "export_value_aud_million",
    "gold_price_usd_oz",
    "gold_price_aud_oz"
]

gold_df[numeric_cols] = gold_df[numeric_cols].apply(pd.to_numeric)
```

```python
gold_df.info()
```

```python
gold_df = gold_df.reset_index(drop=True)
gold_df.head()
```

```python
plt.figure(figsize=(10, 5))
plt.plot(gold_df["date"], gold_df["production_tonnes"])
plt.xlabel("Date")
plt.ylabel("Gold Production (tonnes)")
plt.title("Australian Gold Production Over Time")
plt.show()
```
